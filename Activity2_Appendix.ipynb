```python
import random
import string

LAST_NAME = "BERRE"
SEED_NUM = 2

print("ACTIVITY 2: APPENDIX - ADDITIONAL EXERCISES")
print("Surname:", LAST_NAME)
print("Seed Number:", SEED_NUM)
```

```python
# EXERCISE 1: SENSOR MONITORING

sensor_seed = sum(ord(char) for char in LAST_NAME) + SEED_NUM
rng = random.Random(sensor_seed)

sensor_data = [round(rng.uniform(0, 100), 2) for _ in range(5)]

print("EXERCISE 1: SENSOR MONITORING")
print("Generated Sensor Data:", sensor_data)

valid_readings = []
invalid_readings = []
classification_results = []

# Simulate text-based operator input
operator_inputs = [str(value) for value in sensor_data]
operator_inputs[1] = "invalid"

for index, entry in enumerate(operator_inputs, start=1):
    print(f"\nReading {index}: {entry}")

    try:
        reading = float(entry)

        if not 0 <= reading <= 100:
            invalid_readings.append(entry)
            print("Result: Invalid - outside allowed range")
            continue

        valid_readings.append(reading)

        if reading < 30:
            category = "LOW"
        elif reading <= 70:
            category = "NORMAL"
        else:
            category = "HIGH"

        classification_results.append((reading, category))

        print("Result: Valid")
        print("Classification:", category)

    except ValueError:
        invalid_readings.append(entry)
        print("Result: Invalid - not a number")

print("\n--- EXERCISE 1 SUMMARY ---")
print("Valid Readings:", valid_readings)
print("Invalid Readings:", invalid_readings)
print("Classification Results:", classification_results)
print("Valid Count:", len(valid_readings))
print("Invalid Count:", len(invalid_readings))

if valid_readings:
    average = sum(valid_readings) / len(valid_readings)
    print("Average Valid Reading:", round(average, 2))
else:
    print("Average Valid Reading: No valid readings")

print("Execution Log: Sensor readings processed.")
print("Final Output: Sensor monitoring completed.")
```

```python
# EXERCISE 2: SIGNAL ANALYSIS

signal = f"{LAST_NAME}-{SEED_NUM} Signal! 2026"
processed_signal = signal.strip().upper()

print("EXERCISE 2: SIGNAL ANALYSIS")
print("Generated Signal:", signal)
print("Processed Signal:", processed_signal)

letters = 0
digits = 0
spaces = 0
special_characters = 0

print("\nCharacter Analysis:")

for char in processed_signal:
    if char.isalpha():
        category = "Letter"
        letters += 1
    elif char.isdigit():
        category = "Number"
        digits += 1
    elif char.isspace():
        category = "Space"
        spaces += 1
    else:
        category = "Special Character"
        special_characters += 1

    print(repr(char), "->", category)

if letters > digits:
    signal_classification = "LETTER-DOMINANT"
elif digits > letters:
    signal_classification = "NUMBER-DOMINANT"
else:
    signal_classification = "BALANCED LETTERS AND NUMBERS"

print("\n--- EXERCISE 2 SUMMARY ---")
print("Letters:", letters)
print("Numbers:", digits)
print("Spaces:", spaces)
print("Special Characters:", special_characters)
print("Signal Classification:", signal_classification)
print("Execution Log: Signal normalized and analyzed.")
print("Final Output: Signal analysis completed.")
```

```python
# EXERCISE 3: PASSWORD AUTHENTICATION

password_seed = sum(ord(char) for char in LAST_NAME) + SEED_NUM
password_rng = random.Random(password_seed)

generated_password = (
    LAST_NAME[:3] + str(password_rng.randint(100, 999)) + "!"
)

attempt_limit = 3
attempts_made = 0
access_granted = False
execution_log = []

print("EXERCISE 3: PASSWORD AUTHENTICATION")
print("Generated Password:", generated_password)
print("Attempt Limit:", attempt_limit)

# Demonstration inputs to test authentication
demo_inputs = ["wrong123", "incorrect", generated_password]

for entered_password in demo_inputs:
    if attempts_made >= attempt_limit:
        break

    attempts_made += 1

    if entered_password == generated_password:
        access_granted = True
        execution_log.append(
            f"Attempt {attempts_made}: Access granted"
        )
        print(f"Attempt {attempts_made}: Access granted")
        break
    else:
        execution_log.append(
            f"Attempt {attempts_made}: Access denied"
        )
        print(f"Attempt {attempts_made}: Access denied")

final_state = "UNLOCKED" if access_granted else "LOCKED"

print("\n--- EXERCISE 3 SUMMARY ---")
print("Attempts Made:", attempts_made)
print("Access Result:", "GRANTED" if access_granted else "DENIED")
print("Final System State:", final_state)
print("Execution Log:", execution_log)
print("Final Output: Authentication process completed.")
```
